Now, we can define a set of Genes using:

```python
from pyeas._population import Genes

Genes(bounds=(-1,2), number=10)
```

Next, we define the structure of a 'genome' or member my making a list of genes:

```python
member = [
    Genes(bounds=(-1,0), number=2),
    Genes(bounds=(4,8), number=1),
    Genes(bounds=(0,1), number=2),
]
```

```python
from pyeas._population import Population


pop = Population(
    size=3,
    member=member,
)
```

```python
pop.population_raw
```

```python
pop.groupings
```

```python
pop.bounds
```

```python
pop.population_raw
```

```python
a = pop.denormalise(pop.population_raw)

a
```

```python
pop.normalise(a)
```

```python
d = pop.group(
    pop.denormalise(pop.population_raw)
)
d
```

```python
import numpy as np 
np.shape(d)
```

```python
np.concat(d, axis=0)
```

```python
pop.ungroup(d)
```

```python
pop.normalise(
    pop.ungroup(d)
)
```
